```python
X = asia_df[['vehicle_weight', 'model_year']].iloc[:7,:]
X = X.to_numpy()
X
```

```python
XTX = X@X.T
XTX
```

```python
XTX_inv = np.linalg.pinv(XTX)
XTX_inv
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```
